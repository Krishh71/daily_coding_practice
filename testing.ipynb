```python
df["age"]= df["age"].fillna(df["age"].median())
df
```

```python
df["study_hours"]= df["study_hours"].fillna(df["study_hours"].mean())
```

```python
df["marks"]= df["marks"].fillna(df["marks"].mean())
```

```python
df["city"]= df["city"].fillna(df["city"].mode()[0])
```
